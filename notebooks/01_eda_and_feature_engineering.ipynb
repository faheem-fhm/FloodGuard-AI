# FloodGuard AI: Exploratory Data Analysis & Feature Engineering
This notebook demonstrates exploratory analysis, spatial distribution of elevation across Chennai wards, correlation with flood occurrence, and feature engineering verification.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import json

# Load historical records
df = pd.read_csv('../data/raw/weather_flood_chennai.csv')
print(f"Total records: {len(df)}")
print(df.head())
```

```python
# Class distribution
flood_counts = df['flood_label'].value_counts()
print("Flood occurrence counts:")
print(flood_counts)
print(f"Positive Class Imbalance: {flood_counts[1] / len(df) * 100:.2f}%")
```

```python
# Inspect flood susceptibility by ward
ward_summary = df.groupby('area_name').agg({
    'elevation_m': 'mean',
    'flood_label': ['count', 'sum', 'mean']
})
print(ward_summary)
```
